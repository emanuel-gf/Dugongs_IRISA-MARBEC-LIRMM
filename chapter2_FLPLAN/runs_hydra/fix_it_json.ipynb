```python
def rename_key(obj, old_key="centroid", new_key="aclr"):
    if isinstance(obj, dict):
        new_dict = {}
        for k, v in obj.items():
            k = new_key if k == old_key else k
            new_dict[k] = rename_key(v, old_key, new_key)
        return new_dict

    elif isinstance(obj, list):
        return [rename_key(item, old_key, new_key) for item in obj]

    return obj
```
