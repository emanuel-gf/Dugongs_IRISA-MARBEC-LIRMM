## Compute embeddings

```python
import fiftyone.brain as fob
```

```python
## compute embeddings
results = fob.compute_visualization(dataset,
                                    embeddings=whole_embeddings,
                                    seed=51,
                                    create_index=True,
                                    method = 'tsne',
                                    brain_key='tsne_full_emb',
                                    progress=True,
                                    max_iters = 2000,
                                   learning_rate = 200,
                                   )
```

```python
session = fo.launch_app(dataset,auto=False)
```

```python
session.open_tab()
```

```python
type(results)
```

```python
results.
```

```python
# Visualize embeddings, colored by ground truth label
plot = results.visualize(labels="region")
plot.show(height=720)

session.plots.attach(plot)
```

## UMAP

```python
res_umap = fob.compute_visualization(
    dataset,
    embeddings=whole_embeddings,
    method="umap",
    brain_key="dinov2_class",
    progress=True,
    create_index=True,
    num_neighbors = 15, #neighboring points used in local approximations of manifold structure
    metric='euclidean',
    min_dist = 0.1, # effective minimum distance between embedded points. This controls how tightly the embedding is allowed compress points together
    )
```

```python
# Visualize embeddings, colored by ground truth label
plot = res_umap.visualize(labels="region")
plot.show(height=720)

session.plots.attach(plot)
```
