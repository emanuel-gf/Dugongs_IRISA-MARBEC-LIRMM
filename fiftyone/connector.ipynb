```python
for sample in dataset.iter_samples(progress=True):
    mission_name = sample['mission_name']

    if sample.ground_truth is not None:
        for detection in sample.ground_truth.detections:
            detection["mission_name"] = mission_name
            
        sample.save()
        
```
