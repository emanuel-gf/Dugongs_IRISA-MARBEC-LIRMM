```python
## Patch embeedings - only the bbox of each image
## compute embeedings of image patches
# rather than one embedding per image.
# 1. Define how your model should "embed" a batch of images
import fiftyone.utils.torch as fout
from transformers import AutoImageProcessor

# 1. Setup the processor (crucial for DINOv2)
processor = AutoImageProcessor.from_pretrained("facebook/dinov2-small-imagenet1k-1-layer")

class DINOEmbedder(fout.TorchImageModel):
    def __init__(self, hf_model, processor):
        # We pass a config that tells FiftyOne how to handle the data
        config = fout.TorchImageModelConfig(
            {
                "preprocessing_fcn": None, # We'll handle this in the forward/embed
                "ragged_batches": True,    # Images might be different sizes
                "use_gpu": torch.cuda.is_available(),
            }
        )
        super().__init__(config)
        self.hf_model = hf_model.to(self.device)
        self.processor = processor

    @property
    def has_embeddings(self):
        return True

    def embed(self, arg):
        # FiftyOne calls this. 'arg' is a batch of images (PIL or NumPy)
        with torch.no_grad():
            inputs = self.processor(images=arg, return_tensors="pt").to(self.device)
            outputs = self.hf_model(**inputs, output_hidden_states=True)

            # Extract [CLS] token from the last hidden state
            # Shape: [batch_size, hidden_dim]
            embeddings = outputs.hidden_states[-1][:, 0, :].cpu().numpy()
            return embeddings


# 2. Instantiate and Run
wrapper = DINOEmbedder(model, processor)

# This will now work without the 'BatchFeature' error
patch_embeddings = dataset.compute_patch_embeddings(
    wrapper,
    "ground_truth",
    embeddings_field="patch_embeddings",
    batch_size=16,
    progress=True
)
```

```python

from transformers import AutoImageProcessor

# 1. Setup the processor (crucial for DINOv2)
processor = AutoImageProcessor.from_pretrained("facebook/dinov2-small-imagenet1k-1-layer")

class DINOEmbedder(fo.Model):
    def __init__(self, hf_model, processor):
        self.hf_model = hf_model.to("cuda" if torch.cuda.is_available() else "cpu")
        self.processor = processor

    @property
    def media_type(self):
        return "image"

    @property
    def has_embeddings(self):
        return True

    @property
    def ragged_batches(self):
        return True

    def embed(self, img):
        with torch.no_grad():
            inputs = self.processor(images=img, return_tensors="pt").to(self.hf_model.device)
            outputs = self.hf_model(**inputs, output_hidden_states=True)

            # 1. Get the [CLS] token embeddings
            # Shape is [batch_size, embedding_dim]
            embeddings = outputs.hidden_states[-1][:, 0, :].cpu().numpy()

            # 2. FIX: If FiftyOne passed a single image, return a 1D array.
            # If it passed a list, return a 2D array (which FiftyOne will split).
            if not isinstance(img, list) and embeddings.shape[0] == 1:
                return embeddings.flatten()

            return embeddings

# 1. Instantiate the wrapper
wrapper = DINOEmbedder(model, processor)

# 2. Run patch embeddings (this should now pass the type check)
patch_embeddings = dataset.compute_patch_embeddings(
    wrapper,
    "ground_truth",
    embeddings_field="patch_embeddings",
    batch_size=16,
)
```

```python
## save patch embeddings
# Containers for our data
all_embeddings = []
all_labels = []
all_sample_ids = []
all_detection_ids = []

print("Extracting patch data...")

# We iterate only over samples that have ground_truth detections
for sample in dataset.exists("ground_truth"):
    for det in sample.ground_truth.detections:
        # Important: only grab if the embedding actually exists
        # Note: field name depends on what you used in compute_patch_embeddings
        emb = det.get_field("patch_embeddings")

        if emb is not None:
            all_embeddings.append(emb)
            all_labels.append(det.label)
            all_sample_ids.append(sample.id)
            all_detection_ids.append(det.id)

# Convert to numpy arrays
embeddings_matrix = np.array(all_embeddings)
labels_array = np.array(all_labels)
sample_ids_array = np.array(all_sample_ids)

print(f"Successfully extracted {len(all_embeddings)} patches.")
print(f"Embedding shape: {embeddings_matrix.shape}")

# Save to disk
np.save("dugong_patch_embeddings.npy", embeddings_matrix)
np.save("dugong_patch_labels.npy", labels_array)
np.save("dugong_patch_ids.npy", sample_ids_array)

# Download to your computer
for fname in ["dugong_patch_embeddings.npy", "dugong_patch_labels.npy", "dugong_patch_ids.npy"]:
    files.download(fname)
```

## Detecting Missing Classes and Coverage Gaps

Based on our class distribution analysis, we may have identified underrepresented object classes. Let's systematically find which classes are missing or underrepresented:

```python
# Get all classes present in the dataset
class_counts = dataset.count_values("ground_truth.detections.label")
all_classes = sorted(class_counts.keys())

# Define a threshold for rare classes
threshold = 5  # Consider classes with fewer than 5 instances as rare

# Identify rare classes
rare_classes = []
for cls, count in class_counts.items():
    if count < threshold:
        rare_classes.append((cls, count))

print(f"Total classes in dataset: {len(all_classes)}")
print(f"\nRare classes (< {threshold} instances):")
for cls, count in sorted(rare_classes, key=lambda x: x[1]):
    print(f"  - {cls}: {count} instances")

print("\nMost common classes:")
for cls, count in sorted(class_counts.items(), key=lambda x: x[1], reverse=True)[:5]:
    print(f"  - {cls}: {count} instances")
```
